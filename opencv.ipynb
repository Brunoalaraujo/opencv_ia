```python
import cv2
import matplotlib.pyplot as plt
# from google.colab.patches import cv2_imshow (colocar quando estiver usando o colab)

# Carregar a imagem
imagem = cv2.imread('img/folha_1.jpg')
print(f"Matriz da imagem: {imagem}")  # Imprime a matriz da imagem

# verificar se a imagem foi carregada corretamente
if imagem is None:
    print("Erro ao carregar a imagem.")
else:
    print(f"Dimensões da imagem: {imagem.shape}")

# Converter de BRG (padrão do OpenCV) para RGB (padrão do Matplotlib)
img_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

# Mostrar a imagem usando Matplotlib
plt.imshow(img_rgb)
plt.axis('off')  # Ocultar os eixos
plt.show()

# Mostra a imgem usando cv2_imshow (colocar quando estiver usando o colab)
# cv2_imshow(imagem)
```

```python
# Criar uma cópia da imagem e salvar a cópia com o nome "copia_folha_1.jpg"
if imagem is not None:
    imagem_copia = imagem.copy()
    cv2.imwrite('img/copia_folha_1.jpg', imagem_copia)
    print("Cópia da imagem salva com sucesso.")
else:
    print("Não foi possível criar a cópia da imagem porque a imagem original não foi carregada.")
```

```python
# Definindo a imagem
fig = plt.figure(figsize=(12, 12))

# Separando apenas o canal vermelho (Red)
ax = fig.add_subplot(1, 3, 3)
ax.set_title('Canal Vermelho (Red)')
ax.imshow(imagem[:, :, 2], cmap='gray')  # O canal vermelho é o índice 2
ax.set_xticks([])
ax.set_yticks([])

# Separando apenas o canal verde (Green)
ax = fig.add_subplot(1, 3, 2)
ax.set_title('Canal Verde (Green)')
ax.imshow(imagem[:, :, 1], cmap='gray')  # O canal verde é o índice 1
ax.set_xticks([])
ax.set_yticks([])

# Separando apenas o canal azul (Blue)
ax = fig.add_subplot(1, 3, 1)
ax.set_title('Canal Azul (Blue)')
ax.imshow(imagem[:, :, 0], cmap='gray')  # O canal azul é o índice 0
ax.set_xticks([])
ax.set_yticks([])
```

```python
# Alterando cor de pixel específico (exemplo: alterar o pixel na posição (100, 100) para vermelho puro)

if imagem is not None:
    pixel_original = imagem[100, 100] # indexação começa em 0
    print(f"Valor original do pixel (100, 100): {pixel_original}")
    imagem[100, 100] = [0, 0, 255]  # Define o pixel para vermelho puro (BGR)
    plt.imshow(cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB))  # Mostrar a imagem alterada
    print("Pixel alterado para vermelho puro.")
else:
    print("Não foi possível alterar o pixel porque a imagem não foi carregada.")
```

```python
# Definir ROI (Região de interesse) de linhas

if imagem is not None:
    roi = imagem[50:150, 50:150]  # Define a ROI como um quadrado de 100x100 pixels
    print(f"ROI definida com sucesso. Dimensões da ROI: {roi.shape}")
    plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))  # Mostrar a ROI
    
    # Alterar a ROI para vermelho puro
    imagem[50:150, 50:150] = [0, 0, 255]  # Define a ROI para vermelho puro (BGR)
    plt.imshow(cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB))  # Mostrar a imagem com a ROI alterada
    print("ROI alterada para vermelho puro.")
else:
    print("Não foi possível definir a ROI porque a imagem não foi carregada.")
```
